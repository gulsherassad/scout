```python
import requests, pandas as pd

HEADERS = {"X-Requested-With": "XMLHttpRequest", "User-Agent": "Mozilla/5.0"}
BASE = "https://understat.com"

# Season totals for every EPL player in 2025/26
r = requests.get(f"{BASE}/getLeagueData/EPL/2025", headers=HEADERS, timeout=30)
print(r.status_code)          # should be 200
r.raise_for_status()
league = r.json()
print(league.keys())

players = pd.DataFrame(league["players"])
fixtures = pd.DataFrame(league["dates"])

# Every shot from one finished match
mid = fixtures.loc[fixtures["isResult"] == True, "id"].iloc[0]
m = requests.get(f"{BASE}/getMatchData/{mid}", headers=HEADERS, timeout=30).json()
shots = pd.DataFrame(m["shots"]["h"] + m["shots"]["a"])
print(mid, len(shots), "shots")
```

```python
rost = pd.DataFrame(list(m["rosters"]["h"].values()) + list(m["rosters"]["a"].values()))
print(rost.columns.tolist())
print(rost[["player", "position", "time"]].head(25))
print(rost["position"].value_counts())
```
